```python
cols = df.columns
count = 0
for col in cols:
    if df[col].isna().sum() > 0:
        count += 1
print(count)
```

```python
df_asia = df[df['origin']=='Asia']
df_asia.describe()
```

```python
median_hp = df['horsepower'].median()
most_freq_hp = df['horsepower'].mode()
df_mod = df.copy()
df_mod['horsepower'] = df_mod['horsepower'].fillna(most_freq_hp)
new_median_hp = df_mod['horsepower'].median()
print(f"new median: {new_median_hp}, old median: {median_hp}")
```

```python
import numpy as np
asia_vehicle = df_asia[['vehicle_weight', 'model_year']][:7]
X = np.array(asia_vehicle)
print(X)
```

```python
XTX = np.matmul(X.T, X)
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = (XTX_inv @ X.T) @ y
w
```

```python
sum_w = w.sum()
print(f"sum = {sum_w}")
```
